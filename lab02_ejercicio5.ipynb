```python
numeros = [1, 2, 3, 4, 5, 6, 7, 8]

pares = filter(lambda x: x % 2 == 0, numeros)

resultado = list(map(lambda x: x ** 2, pares))

print(resultado)
```

```python
numeros = [2, 3, 4, 5, 6]

cubos = map(lambda x: x ** 3, numeros)

mayores = filter(lambda x: x > 100, cubos)

resultado = list(map(lambda x: str(x), mayores))

print(resultado)
```

```python
numeros = [5, 12, 3, 20, 15, 8]

mayores = filter(lambda x: x > 10, numeros)

cuadrados = map(lambda x: x ** 2, mayores)

resultado = sorted(cuadrados, reverse=True)

print(resultado)
```
