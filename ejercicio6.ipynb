```python
def minimax(piedras, es_max):

    # Si ya no quedan piedras
    if piedras == 0:
        if es_max:
            return -1
        else:
            return 1

    if es_max:

        mejor = -1000

        for quitar in [1, 2, 3]:

            if piedras >= quitar:
                valor = minimax(piedras - quitar, False)
                mejor = max(mejor, valor)

        return mejor

    else:

        peor = 1000

        for quitar in [1, 2, 3]:

            if piedras >= quitar:
                valor = minimax(piedras - quitar, True)
                peor = min(peor, valor)

        return peor


def mejor_movimiento(piedras):

    mejor_valor = -1000
    movimiento = 1

    for quitar in [1, 2, 3]:

        if piedras >= quitar:

            valor = minimax(piedras - quitar, False)

            if valor > mejor_valor:
                mejor_valor = valor
                movimiento = quitar

    return movimiento


# Programa principal

piedras = 10

print("=== JUEGO NIM ===")

while piedras > 0:

    print("\nPiedras restantes:", piedras)

    # TURNO HUMANO (MIN)
    quitar = int(input("¿Cuántas piedras deseas quitar (1-3)? "))

    while quitar not in [1, 2, 3] or quitar > piedras:
        quitar = int(input("Movimiento inválido. Ingrese 1, 2 o 3: "))

    piedras -= quitar

    if piedras == 0:
        print("\nNo quedan piedras.")
        print("Gana la computadora.")
        break

    print("\nLa computadora está pensando...")

    # TURNO COMPUTADORA (MAX)
    movimiento_cpu = mejor_movimiento(piedras)

    print("La computadora quita:", movimiento_cpu)

    piedras -= movimiento_cpu

    if piedras == 0:
        print("\nNo quedan piedras.")
        print("Gana el jugador humano.")
        break
```
