```python
from functools import reduce

numeros = [2, 3, 4]

resultado = reduce(
    lambda a, b: a * b,
    numeros
)

print(resultado)
```

```python
from functools import reduce

palabras = ["Hola", " ", "Mundo"]

resultado = reduce(
    lambda a, b: a + b,
    palabras
)

print(resultado)
```

```python
from functools import reduce

numeros = [5, 12, 8, 20, 3]

resultado = reduce(
    lambda a, b: a if a > b else b,
    numeros
)

print(resultado)
```
