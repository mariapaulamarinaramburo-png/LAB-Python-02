```python
palabras = ["pera", "manzana", "uva", "banana"]

resultado = sorted(palabras)

print(resultado)
```

```python
tuplas = [
    ("Carlos", 25),
    ("Ana", 19),
    ("Pedro", 30),
    ("Juan", 20)
]

resultado = sorted(tuplas, key=lambda x: x[0])

print(resultado)
```

```python
productos = [
    ("Camisa", 30),
    ("Pantalón", 55),
    ("Medias", 10),
    ("Chaqueta", 80)
]

resultado = sorted(
    productos,
    key=lambda x: (-x[1], x[0])
)

print(resultado)
```

```python
palabras = ["sol", "computador", "mesa", "gato", "universidad"]

resultado = sorted(
    palabras,
    key=lambda x: len(x)
)

print(resultado)
```

```python
numeros = [7, 2, 9, 4, 1, 8, 6, 3]

resultado = sorted(
    numeros,
    key=lambda x: x % 2
)

print(resultado)
```

```python
palabra = "programacion"

frecuencia = {}

for letra in palabra:
    frecuencia[letra] = frecuencia.get(letra, 0) + 1

resultado = sorted(
    frecuencia.items(),
    key=lambda x: x[1],
    reverse=True
)

print(resultado)
```

```python
empleados = [
    ("Maria", "Ventas", 30),
    ("Luis", "Tecnología", 25),
    ("Ana", "Ventas", 25),
    ("Pedro", "Tecnología", 28)
]

resultado = sorted(
    empleados,
    key=lambda x: (x[1], x[2])
)

print(resultado)
```
