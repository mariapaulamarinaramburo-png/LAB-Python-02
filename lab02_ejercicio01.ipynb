```python
multiplo3 = lambda numero: numero % 3 == 0

print(multiplo3(9))
print(multiplo3(10))
```

```python
cubo = lambda numero: numero ** 3

print(cubo(2))
print(cubo(5))
```

```python
producto = lambda a, b: a * b

print(producto(4, 5))
```

```python
mayor = lambda a, b: a if a > b else b

print(mayor(8, 3))
```

```python
empiezaA = lambda palabra: palabra.startswith(("A", "a"))

print(empiezaA("Ana"))
print(empiezaA("casa"))
```

```python
fahrenheit = lambda c: (c * 9/5) + 32

print(fahrenheit(25))
```
