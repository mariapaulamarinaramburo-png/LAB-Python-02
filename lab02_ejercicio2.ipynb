```python
funciones = [
    lambda x: x * 2,
    lambda x: x + 10,
    lambda x: x ** 2
]

numero = int(input("Ingrese un número: "))

for funcion in funciones:
    print(funcion(numero))
```

```python
palabras = ["sol", "perro", "gato", "computador", "mesa"]

resultado = list(filter(lambda x: len(x) > 4, palabras))

print(resultado)
```

```python
lista = [10, None, 5, None, 8]

resultado = list(filter(lambda x: x is not None, lista))

print(resultado)
```

```python
palabras = ["Ana", "casa", "uva", "elefante", "perro"]

resultado = list(filter(
    lambda x: x.lower().startswith(("a","e","i","o","u")),
    palabras
))

print(resultado)
```

```python
palabras = ["oso", "ana", "casa", "reconocer"]

resultado = list(filter(
    lambda x: x == x[::-1],
    palabras
))

print(resultado)
```

```python
numeros = [15, 22, 35, 40, 55]

resultado = list(filter(
    lambda x: x % 10 == 5,
    numeros
))

print(resultado)
```

```python
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]

resultado = list(filter(
    lambda producto: producto["precio"] > 100,
    productos
))

print(resultado)
```
